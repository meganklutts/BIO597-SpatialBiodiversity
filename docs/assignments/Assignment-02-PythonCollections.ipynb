## 1. Create a list

A list stores multiple values in one variable. Lists use square brackets.

Create a list called `snake_species` with at least four species names as strings.

```python
snake_species = [
    "Storeria dekayi",
    "Agkistrodon contortrix",
    "Pantherophis guttatus",
    "Diadophis punctatus",
]

snake_species
```

## 2. Access items by position

Python lists are zero-indexed. This means the first item is at position `0`, the second item is at position `1`, and so on.

Fill in the blanks to print the first and third species.

```python
print(snake_species[0])
print(snake_species[2])
```

## 3. Access the last item

Negative indexing counts from the end of a list. The last item is at position `-1`.

Print the last species in your list.

```python
# Write one line of code here.
print(snake_species[3])
```

## 4. Count the number of items

`len()` tells you how many items are in a list.

Use `len()` to count how many species are in `snake_species`.

```python
number_of_species = len(snake_species)
number_of_species
```

## 5. Add a new item to a list

`append()` adds one item to the end of a list.

Add one more snake species to `snake_species`, then display the updated list.

```python
snake_species.append("Agkistrodon contortrix")
snake_species
```

## 6. Sort a list

`sorted()` returns a sorted version of a list. It does not permanently change the original list.

Sort the species names alphabetically.

```python
sorted_species = sorted(snake_species)
sorted_species
```

## 7. Loop over a list

A `for` loop lets you do the same thing to every item in a list.

Fill in the blank so the loop prints each species name.

```python
for species_name in snake_species:
    print(species_name)
```

## 8. Use list values inside a sentence

Inside a loop, you can use each list item to build a message.

Complete the loop so it prints a sentence for each species.

```python
for species_name in snake_species:
    print("GBIF can be searched for " + species_name)
```

## 9. Create a dictionary

A dictionary stores key-value pairs. Dictionaries use curly braces.

Create a dictionary for one species. The keys are already provided. Fill in the missing values.

```python
dekayi_record = {
    "scientificName": "Storeria dekayi",
    "commonName": "DeKay's brownsnake",
    "family": "Colubridae",
    "hasCoordinates": True,
}

dekayi_record
```

## 10. Access values by key

Dictionary values are accessed by key, not by position.

Print the scientific name and common name from `dekayi_record`.

```python
print(dekayi_record["scientificName"])
print(dekayi_record["commonName"])
```

## 11. Look at dictionary keys

`.keys()` shows the keys available in a dictionary. This is useful when you are exploring unfamiliar data.

Use `.keys()` on `dekayi_record`.

```python
dekayi_record.keys()
```

## 12. Update a dictionary value

Dictionaries are mutable, which means you can change their values.

Change `hasCoordinates` from `True` to `False`, then display the dictionary.

```python
dekayi_record["hasCoordinates"] = False
dekayi_record
```

## 13. Add a new key-value pair

You can add a new key to a dictionary by assigning a value to it.

Add a new key called `source` with the value `GBIF`.

```python
dekayi_record["source"] = "GBIF"
dekayi_record
```

## 14. Loop over a dictionary

When you loop over a dictionary, Python gives you each key. You can use that key to access the value.

Complete the loop so it prints each key and its value.

```python
for key in dekayi_record:
    print(key, dekayi_record[key])
```

## 15. Create a list of dictionaries

Many API results are structured like this: a list where each item is a dictionary.

The example below is a small fake version of that structure. Fill in the missing species name in the third dictionary.

```python
occurrence_records = [
    {"scientificName": "Storeria dekayi", "year": 2020, "countryCode": "US"},
    {"scientificName": "Agkistrodon contortrix", "year": 2018, "countryCode": "US"},
    {"scientificName": "Pantherophis guttatus", "year": 2022, "countryCode": "US"},
]

occurrence_records
```

## 16. Access a dictionary inside a list

First choose an item from the list by position. Then choose a value from that dictionary by key.

Fill in the blanks to print the scientific name from the second record.

```python
second_record = occurrence_records[1]
second_record["scientificName"]
```

## 17. Loop over a list of dictionaries

This pattern is very common when working with API data.

Complete the loop so it prints the scientific name from each record.

```python
for record in occurrence_records:
    print(record["scientificName"])
```

## 18. Use an `if` statement inside a loop

Loops can be combined with `if` statements to select records that match a condition.

Complete the code so it prints only records from 2020 or later.

```python
for record in occurrence_records:
    if record["year"] >= 2020:
        print(record)
```

## 19. Build a new list inside a loop

Sometimes we want to collect selected values into a new list.

The cell below creates an empty list, then appends species names from each record.

```python
names_from_records = []

for record in occurrence_records:
    names_from_records.append(record["scientificName"])

names_from_records
```
